# Purpose
This file uses the given inventory of kitchens in "building_kitchen_list.csv" to produce a list of kitchens with condensed features ("condensed_kitchens.csv"), a list of residential kitchens ("residential_kitchens.csv"), and a list of public kitchens ("public_kitchens.csv"). The list of public kitchens is used in conjunction with the list of existing water-filling stations in our calculation for the number of people per water-filling-station in the file "../foot-traffic/Estimated foot traffic ultimate.ipynb".

```python
import pandas as pd
```

```python
kitchens = pd.read_csv("./building_kitchen_list.csv")

# strip extra whitespaces
kitchens = kitchens.map(lambda x: x.strip() if isinstance(x, str) else x)

# filter the useful columns
condensed_kitchens = kitchens[["Building Code (BLDGCODE)", "Building Description (BLDGDESC)", "Building Type (BLDGTYPE)", "Floor (FLOORCODE)", "Room # (RMID)"]]
condensed_kitchens.to_csv("condensed_kitchens.csv", index=False)
condensed_kitchens.head()
```

```python
# counts the null values in the kitchen data
na_counts = kitchens.isna().sum()
print(na_counts)
```

```python
# filters residential kitchens
residential_kitchens = kitchens[kitchens["Building Type (BLDGTYPE)"] == "Residential"]
residential_kitchens.to_csv("residential_kitchens.csv", index=False)
print(residential_kitchens.head())
print(f"{len(residential_kitchens)} residential kitchens")
```

```python
# filters kitchens in Warren Towers, Sleeper Hall, Rich Hall, Claflin Hall, and Bay State Road Brownstones
public_kitchens = kitchens.copy()

def is_public_kitchen(row):
    # Warren Towers
    if row["Building Description (BLDGDESC)"] == "700 COMMONWEALTH AVENUE":
        return 1
    # Claflin Hall
    elif row["Building Description (BLDGDESC)"] == "273 BABCOCK STREET":
        return 1
    # Sleeper Hall
    elif row["Building Description (BLDGDESC)"] == "275 BABCOCK STREET":
        return 1
    # Rich Hall
    elif row["Building Description (BLDGDESC)"] == "277 BABCOCK STREET":
        return 1
    # Bay State Road Residences
    elif "BAY STATE ROAD" in row["Building Description (BLDGDESC)"] and row["Building Type (BLDGTYPE)"] == "Residential":
        return 1
    else:
        return 0

public_kitchens["Public Kitchen"] = public_kitchens.apply(is_public_kitchen, axis=1)

public_kitchens = public_kitchens[public_kitchens["Public Kitchen"] == 1]
public_kitchens.to_csv("public_kitchens.csv", index=False)
print(public_kitchens.head())
print(f"{len(public_kitchens)} public kitchens")
```
