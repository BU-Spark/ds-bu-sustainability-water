
In our analysis of the impact of mechanical ventilation, we noted a significant increase in both average and maximum densities at locations fully equipped with mechanical ventilation systems. Therefore, we focused on these locations to determine the necessity of adding new water bottle stations. 

then we select the candidates for water bottle staion using the following standard.

1. remain all high volume places no matter they fit 2 and 3 condition or not.

2. consider the place where mechanical ventilation is 'yes'

3. set 75% of average mean and max density of covered floors as mean and max density threshold. 

and select which max and mean density are larger than thresold

```python
import pandas as pd

# read data
data_covered = pd.read_csv('covered_floors.csv')
data_minus_covered = pd.read_csv('uncovered_floors.csv')

# calculate average mean and max density for covered places
covered_mean_density_avg = data_covered['mean_density_cnt'].mean()
covered_max_density_avg = data_covered['max_density_cnt'].mean()

# define threhold as 75% of average mean and max density of covered floor
threshold_mean_density = covered_mean_density_avg * 0.75
threshold_max_density = covered_max_density_avg * 0.75

# select candidates as what 
candidates = data_minus_covered[
    ((data_minus_covered['mean_density_cnt'] >= threshold_mean_density) | 
     (data_minus_covered['max_density_cnt'] >= threshold_max_density) |
     (data_minus_covered['high_volumn'] == 'TRUE')) &  # Assuming 'high_volume' column exists and is Boolean
    (data_minus_covered['Mechanical Ventilation'] == 'Yes')
]

# print result
candidates.to_csv('selected_candidates.csv', index=False)  # Saving to a CSV file without the index
print(candidates)
```

```python
df = candidates

colors = {
    'Academic': 'blue',
    'Athletic': 'green',
    'Research': 'red',
    'Residential': 'orange',
    'Student Support': 'purple'
}

markers = {
    'Academic': 'o',
    'Athletic': '^',
    'Research': 's',
    'Residential': 'D',
    'Student Support': 'P'
}

plt.figure(figsize=(8, 6))

for building_type in df['building_type'].unique():
    subset = df[df['building_type'] == building_type]
    plt.scatter(subset['longitude'], subset['latitude'], 
                s=subset['mean_density_cnt']*5,  # Scaling for better visualization
                c=colors[building_type], 
                marker=markers[building_type], 
                alpha=0.7, 
                label=building_type)

plt.xlabel('Longitude', fontweight='bold')
plt.ylabel('Latitude', fontweight='bold')
plt.title('Location Plot of Candidates with Marker Size Proportional to Mean Density Count', fontweight='bold')
plt.legend(title='Building Type', loc='upper right', prop={'size': 15})
plt.grid(True)

#plt.show()
plt.savefig('location of candidates.png')
```
