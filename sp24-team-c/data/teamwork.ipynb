```python
import pandas as pd
```

```python
building_and_kitchen = pd.read_csv("Building and Kitchen List.csv")
floors = pd.read_csv("foot_traffic_minus_covered_floors.csv")
```

```python
building_and_kitchen.head()
```

```python
floors.head()
```

```python
# remove rows whose building_floor is 'p' because we don't need parking

floors_filtered = floors[floors['building_floor'] != 'p']
```

```python
# merge two dataframes using left outer join

merged_df = pd.merge(floors_filtered, building_and_kitchen, how = 'left', left_on = 'building_desc', right_on = 'Building Description')
```

```python
merged_df.head()
```

```python
# add a new column：'Maintained by BU FMO'

final_columns = list(floors_filtered.columns)
final_columns.append('Maintained by BU FMO')
```

```python
final_df = merged_df[final_columns]
```

```python
final_df.to_csv('foot_traffic_minus_covered_floors_processed.csv', index=False)
```
