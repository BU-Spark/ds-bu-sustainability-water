Geospatial Using Folium: 
Our main goal for this research is to determine where new bottle filling stations should be located, not where they currently exist. By utilizing the different datasets (wifi, kitchen, density, etc.), we can determine the which building need bottle filling stations.

```python
import folium
from IPython.display import display

m = folium.Map(location=(42.3505, -71.1054), tiles="cartodb positron")

folium.Marker(
    location=[42.3505, -71.1054],
    tooltip="Click me!",
    popup="BU",
    icon=folium.Icon(icon="cloud"),
).add_to(m)

display(m)
```
