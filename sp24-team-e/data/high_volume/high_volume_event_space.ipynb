# This is the Python file that analyzes and cleans the dataset with buildings that had high volume event space

```python
import pandas as pd

df = pd.read_csv('High Volume Event Spaces.csv')
# df.columns = df.iloc[0]
# df = df.drop(0)
df.head(10)
```

Since the dataset is small, made some adjustments through fixing the csv file itself. <br>
- Deleted the first row with just a title of the dataset and shifted up the entire dataset by one row
- Got rid of a column with nothing in it.
- For "Current," "-1" means we do not know for sure how many there are. It was previously "??"
- Added a column "Candidacy" which represents whether the place is a good candidate for water station or not. In the original file it was indicated by color of the cell.
    - <b>Good</b>: "Good candidate for bottle fillers." Represented by Green
    - <b>Maybe</b>: "Maybe." Represented by Orange
    - <b>Bad</b>: "Not a good candidate." Represented by Red
    - <b>Enough</b>: "Already completed/sufficient." Represented by Gray
    - <b>More_research_need</b>: Added this label for cells that did not have any color.
- Got rid of the legends note at the bottom of the dataset.
- Added a column name "Installed Time" because there was no title describing the column.

### Current number of water station
1. If NaN, it is -99
2. If not sure, it is -1
3. Rest of them are equivalent to their actual number. If there is 1, it is 1.

```python
replacement_value_current = -99
df['Current'] = df['Current'].fillna(replacement_value_current)
df.head(10)
```

```python
df
```

### Installed Time of water station
- If it is NaN, it is replaced by -99.

```python

replacement_value_installed_time = -99
df['Installed_Time'] = df['Installed_Time'].fillna(replacement_value_installed_time)
df
```

### Replacing Candidacy with dummy variables
1. Good = 1
2. Maybe = 2
3. Bad = 3
4. Enough = 4
5. More_research_needed = 5

```python
mapping = {'Good': 1, 'Maybe': 2, 'Bad': 3, 'Enough': 4, 'More_research_needed': 5}

df['Candidacy'] = df['Candidacy'].map(mapping)
df
```
